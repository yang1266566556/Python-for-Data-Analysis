```python
numpy_data_7 = np.arange(0, 9, 1)

np.random.shuffle(numpy_data_7)

numpy_data_7 = np.reshape(numpy_data_7, (3, 3))

print(numpy_data_7)

print(np.argmin(numpy_data_7))
print(np.argmax(numpy_data_7))
print(np.argwhere(numpy_data_7 > 7))

numpy_data_7[numpy_data_7 > 6] = 10
print(numpy_data_7)

numpy_data_rank = np.argsort(numpy_data_7)
print(numpy_data_rank)
```

```python
'''
| 代码                        | 作用 |
|-----------------------------|------------------------------|
| `np.arange(0,9,1)`          | 生成 0 到 8 |
| `np.random.shuffle()`       | 随机打乱 |
| `np.reshape(..., (3,3))`    | 改成 3×3 矩阵 |
| `np.argmin()`               | 返回最小值的位置 |
| `np.argmax()`               | 返回最大值的位置 |这两个函数都会先转化为一维才返回索引
| `np.argwhere(arr > 7)`      | 返回满足条件的位置 |
| `arr[arr > 6] = 10`         | 把大于 6 的数改成 10 |
| `np.argsort()`              | 返回排序后的索引 |
'''
```

```python
'''
| 函数                    | 返回内容           | 默认行为 |
|---------------------    |---=--------------- |-----------------|
| `np.argmax(a)`          | 最大值的位置       | 返回展开后的一维索引 |
| `np.argmin(a)`          | 最小值的位置       | 返回展开后的一维索引 |
| `np.argwhere(a > 条件)` | 所有满足条件的位置 | 返回原数组'坐标' |即二维数组就返回二维坐标三维返回三维坐标其余同理
'''
```

```python
'''
课堂练习：

• 使用Pandas读取Excel文件

• 将Pandas转换为Numpy矩阵

• 寻找该矩阵中第3大的值所在的位置

• 将该矩阵中大于75%分位数的值修改为最大值

• 保存该矩阵为Excel文件

• 通过阅读Numpy文档，每个同学调用3个本节课未讲述的Numpy方法，并进行解释
'''
```

```python
excel_data = pd.read_excel('Unit2_result.xlsx', index_col=0)
```

```python
# 2. 将 Pandas 转换为 Numpy 矩阵
numpy_matrix = excel_data.to_numpy()
print("转换后的 Numpy 矩阵：")
print(numpy_matrix)
```

```python
#3寻找该矩阵中第3大的值所在的位置
third_largest_value = np.sort(numpy_matrix.flatten())[-3]

third_largest_position = np.argwhere(numpy_matrix == third_largest_value)

print("第3大的值是：", third_largest_value)
print("第3大的值所在位置：", third_largest_position)
```

```python
# 4. 将该矩阵中大于75%分位数的值修改为最大值
q75 = np.quantile(numpy_matrix, 0.75)
max_value = np.max(numpy_matrix)
print("75%分位数是：")
print(q75)
print("最大值是：")
print(max_value)
numpy_matrix[numpy_matrix > q75] = max_value
print("修改后的矩阵：")
print(numpy_matrix)
```

```python
# 5. 保存该矩阵为 Excel 文件
result_data = pd.DataFrame(numpy_matrix, columns=excel_data.columns)
result_data.to_excel('Unit2_final_result.xlsx', index=False)
print("修改后的矩阵已保存为 Unit2_final_result.xlsx")
```
