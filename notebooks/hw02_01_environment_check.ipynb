## 环境、依赖版本与临时文件读写检查

```python
import sys
import json
import platform
import tempfile
from pathlib import Path
from importlib.metadata import version, PackageNotFoundError

# 与第一次作业相同：定位专题目录，再使用公共初始化。
# 支持从仓库根目录或 notebooks/ 启动Jupyter，不限定安装盘符。
search_start = Path.cwd().resolve()
for candidate in (search_start, *search_start.parents):
    notebook_dir = candidate if (candidate / "railway_analysis/common.py").is_file() else candidate / "notebooks"
    if (notebook_dir / "railway_analysis/common.py").is_file():
        if str(notebook_dir) not in sys.path:
            sys.path.insert(0, str(notebook_dir))
        break
else:
    raise FileNotFoundError("未找到 railway_analysis/common.py，请从仓库或 notebooks 目录启动Jupyter。")

from railway_analysis.common import initialize_project

assignment_dir = initialize_project(check_versions=False)

packages = [
    "ipykernel", "numpy", "pandas", "scipy",
    "scikit-learn", "umap-learn", "matplotlib",
    "requests", "networkx", "pyproj",
]

package_versions = {}
for package in packages:
    try:
        package_versions[package] = version(package)
    except PackageNotFoundError:
        package_versions[package] = "未安装或未找到版本信息"

# 实际创建、读取临时文件，避免仅凭路径存在判断权限。
try:
    with tempfile.TemporaryDirectory(
        prefix="railway_check_", dir=assignment_dir
    ) as temporary_dir:
        test_file = Path(temporary_dir) / "读写测试.txt"
        test_file.write_text("铁路分析环境核验", encoding="utf-8")
        assert test_file.read_text(encoding="utf-8") == "铁路分析环境核验"
    file_check = "创建、写入、读取、清理均通过"
except Exception as error:
    file_check = f"{type(error).__name__}: {error}"

diagnosis = {
    "Python版本": platform.python_version(),
    "当前解释器": sys.executable,
    "是否满足Python版本要求": sys.version_info[:2] == (3, 12),
    "当前工作目录": str(Path.cwd()),
    "作业目录": str(assignment_dir),
    "文件读写测试": file_check,
    "依赖版本": package_versions,
}
print(json.dumps(diagnosis, ensure_ascii=False, indent=2))
```
