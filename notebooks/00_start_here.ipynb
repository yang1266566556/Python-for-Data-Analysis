```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "day": pd.date_range("2026-01-01", periods=7),
    "study_hours": [1.0, 1.5, 2.0, 1.2, 2.4, 2.8, 3.0],
})
df
```

```python
sns.lineplot(data=df, x="day", y="study_hours", marker="o")
plt.title("Study Hours")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
