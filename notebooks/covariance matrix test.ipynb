```python
import os
import pandas as pd
import numpy as np

print("当前目录：", os.getcwd())
print("当前目录文件：", os.listdir())
print("pandas版本：", pd.__version__)
```

```python
import pandas as pd
import numpy as np

df = pd.read_excel(
    "大学生素质影响因素表.xlsx",
    engine="openpyxl"
)

display(df)
print("表格行列数：", df.shape)
print("列名：", df.columns.tolist())
```

```python
data = df.iloc[:, 1:5].copy()
data.columns = ["X1", "X2", "X3", "X4"]

display(data)
```

```python
# 转换为数值，无法转换的内容会变成空值
data = data.apply(pd.to_numeric, errors="coerce")

# 删除整行为空的行
data = data.dropna(how="all").reset_index(drop=True)

# 设置学生编号
data.index = range(1, len(data) + 1)
data.index.name = "学生编号"

print("整理后的数据：")
display(data)

print("样本量：", len(data))

print("各列空值数量：")
display(data.isna().sum())
```

```python
# 检查数据是否有问题
if len(data) != 20:
    print("警告：题目样本量应为20，当前读取到", len(data), "行。")

if data.isna().any().any():
    print("以下行存在空值或非数字：")
    display(data[data.isna().any(axis=1)])
    raise ValueError("请检查Excel中的数据。")

n = len(data)

# 1. 样本均值向量
mean_vector = data.mean()

# 2. 样本离差矩阵
A = data - mean_vector

# 离差平方和与交叉乘积矩阵
C = A.T @ A

# 3. 样本协方差矩阵，使用分母 n-1
S = data.cov(ddof=1)

# 4. 样本相关矩阵
R = data.corr()

print("样本量 n =", n)

print("\n1. 样本均值向量：")
display(mean_vector.to_frame().T.round(4))

print("\n2. 样本离差矩阵 A：")
display(A.round(4))

print("\n离差平方和与交叉乘积矩阵 AᵀA：")
display(C.round(4))

print("\n3. 样本协方差矩阵 S：")
display(S.round(4))

print("\n4. 样本相关矩阵 R：")
display(R.round(4))
```

```python
S_check = C / (n - 1)

print("根据公式 AᵀA/(n-1) 计算的协方差矩阵：")
display(S_check.round(4))

print(
    "两种算法是否一致：",
    np.allclose(S.values, S_check.values)
)
```

```python
output_file = "大学生素质影响因素_计算结果.xlsx"

mean_df = pd.DataFrame(
    [mean_vector.values],
    columns=data.columns,
    index=["样本均值"]
)

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    data.to_excel(writer, sheet_name="整理后的原始数据")
    mean_df.to_excel(writer, sheet_name="样本均值向量")
    A.to_excel(writer, sheet_name="样本离差矩阵")
    C.to_excel(writer, sheet_name="离差平方和交叉乘积阵")
    S.to_excel(writer, sheet_name="样本协方差矩阵")
    R.to_excel(writer, sheet_name="样本相关矩阵")

print("保存成功：", output_file)
```

```python
corr_result = pd.read_excel(
    "大学生素质影响因素_计算结果.xlsx",
    sheet_name="样本相关矩阵",
    index_col=0,
    engine="openpyxl"
)

display(corr_result)
```

```python
从 `大学生素质影响因素_计算结果.xlsx` 里可以得到这些结论：

1. **平均影响程度排序**  
   四个因素的均值分别是：X1 = 8.15，X4 = 7.10，X2 = 5.10，X3 = 4.45。  
   说明在这 20 名学生中，**家庭环境与家庭教育 X1 的影响评分最高**，其次是**个人向上发展的心理动机 X4**；学校生活环境 X2 和学校周围环境 X3 的平均影响相对较低。

2. **学生意见分歧最大的是 X3**  
   方差从大到小为：X3 = 4.997，X2 = 3.147，X4 = 2.516，X1 = 1.503。  
   说明大家对**学校周围环境 X3**影响程度的看法差异最大；对**家庭环境与家庭教育 X1**的看法最一致。

3. **X1 与 X2 呈较明显负相关**  
   相关系数 `r = -0.5639`，是所有变量关系里绝对值最大的。  
   可以理解为：在样本中，家庭环境与家庭教育评分越高的学生，往往对学校生活环境的影响评分越低，反之亦然。这个关系相对值得关注。

4. **其他变量之间相关性较弱**  
   X2 与 X3 有一定正相关，`r = 0.3331`；X3 与 X4 为弱正相关，`r = 0.2093`。  
   X2 与 X4 几乎没有线性关系，`r = 0.0337`。  
   因此除了 X1-X2 外，其他因素之间没有表现出很强的线性联系。

总体结论：**家庭环境与家庭教育是样本中被认为影响最大的因素，个人发展动机次之；学校周围环境虽然平均分不高，但个体差异最大。因素之间整体相关性不强，只有家庭环境与学校生活环境之间存在较明显的负向关系。**

需要注意：这些结果来自 20 个样本，只能说明样本中的描述性特征和相关关系，不能直接证明因果关系。
```
