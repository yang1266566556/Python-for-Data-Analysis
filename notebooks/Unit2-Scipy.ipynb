```python
import numpy as np
import scipy.spatial.distance as scipy_distance
from scipy import stats
```

```python

numpy_data_7 = np.arange(0, 100, 1)

numpy_data_9 = np.arange(0, 200, 2)

distance = scipy_distance.euclidean(numpy_data_7, numpy_data_9)
print(distance)#两个矩阵间的欧式距离
```

```python
distance = scipy_distance.cosine(numpy_data_7, numpy_data_9)
print(distance)#余弦距离

distance = scipy_distance.correlation(numpy_data_7, numpy_data_9)
print(distance)#相关距离

distance = scipy_distance.hamming(numpy_data_7, numpy_data_9)
print(distance)#汉明距离 = 两个等长字符串/向量中，对应位置不同的个数,这里课件上实际并非明汉距离为明汉距离比例，0.99即99个数不同，即明汉距离为99
```

```python
'''
课堂练习：

• 利用Numpy，生成2个随机数矢量，元素数量自定，并打印。

• 计算上述矢量的欧氏、余弦、汉明距离。

• 自己再探索1个Scipy的函数，并解释其功能。
'''
```

```python
# 1. 利用 Numpy 生成 2 个随机数矢量，元素数量自定，并打印
# 这里设置元素数量为 10
np.random.seed(0)  # 固定随机种子，保证每次运行结果一样

vector_1 = np.random.randint(0, 10, size=10)
vector_2 = np.random.randint(0, 10, size=10)

print("第一个随机数矢量 vector_1：")
print(vector_1)

print("第二个随机数矢量 vector_2：")
print(vector_2)
```

```python
# 2. 计算上述矢量的欧氏距离、余弦距离、汉明距离
# 欧氏距离
euclidean_distance = scipy_distance.euclidean(vector_1, vector_2)

# 余弦距离
cosine_distance = scipy_distance.cosine(vector_1, vector_2)

# 汉明距离
hamming_distance = np.sum(vector_1 != vector_2)

print("欧氏距离：")
print(euclidean_distance)

print("余弦距离：")
print(cosine_distance)

print("汉明距离：")
print(hamming_distance)
```

```python
'''
| 写法                                    | 结果含义 |
|-----------------------------------------|----------|
| `scipy_distance.hamming(a, b)`          | 不同位置所占比例 |
| `np.sum(a != b)`                        | 不同位置的个数   |
| `scipy_distance.hamming(a, b) * len(a)` | 不同位置的个数 |
'''
```

```python
# 3. 自己再探索 1 个 Scipy 的函数，并解释其功能
# 这里选择 scipy.stats.mode()，用于寻找众数
mode_result = stats.mode(vector_1, keepdims=True)
print("vector_1 的众数结果：")
print(mode_result)
print("vector_1 中出现次数最多的值是：")
print(mode_result.mode)
print("该值出现的次数是：")
print(mode_result.count)
```
